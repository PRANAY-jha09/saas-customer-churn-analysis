# Business Problem

# What is the overall customer churn rate?

```python
# Calculate the overall customer churn rate
total_accounts = len(accounts)
churned_accounts = accounts['churn_flag'].sum()
churn_rate = (churned_accounts / total_accounts) * 100

print(f"Total Accounts: {total_accounts}")
print(f"Churned Accounts: {churned_accounts}")
print(f"Overall Customer Churn Rate: {churn_rate:.2f}%")
```

# Bussiness Insight :- The overall churn rate is 22.0%, meaning customer retention is a clear business risk

# Which plan tiers have the highest churn?

```python
# Calculate churn rate by plan tier
plan_churn = accounts.groupby('plan_tier')['churn_flag'].agg(
    total_accounts='count',
    churned_accounts='sum',
    churn_rate='mean'
).reset_index()

# Convert churn rate to percentage
plan_churn['churn_rate'] = plan_churn['churn_rate'] * 100
plan_churn = plan_churn.sort_values(by='churn_rate', ascending=False)

display(plan_churn)
```

# Bussiness Insight :- plan_tier(Pro) have the highest churn

# Which industries are most at risk?

```python
# Calculate churn rate by industry
industry_churn = accounts.groupby('industry')['churn_flag'].agg(
    total_accounts='count',
    churned_accounts='sum',
    churn_rate='mean'
).reset_index()

# Convert churn rate to percentage
industry_churn['churn_rate'] = industry_churn['churn_rate'] * 100
industry_churn = industry_churn.sort_values(by='churn_rate', ascending=False)

display(industry_churn)
```

# Bussiness_insight :- DEVTOOLS is the most churned out of them its churn_rate is 30.9 .

# Which referral sources bring higher-churn customers?

```python
# Calculate churn rate by referral source
referral_churn = accounts.groupby('referral_source')['churn_flag'].agg(
    total_accounts='count',
    churned_accounts='sum',
    churn_rate='mean'
).reset_index()

# Convert churn rate to percentage and sort
referral_churn['churn_rate'] = referral_churn['churn_rate'] * 100
referral_churn = referral_churn.sort_values(by='churn_rate', ascending=False)

display(referral_churn)
```

# Bussiness insight :- Event referral_source is the bring higher-churn customers.

# Which customer seat-size groups churn the most?


```python
# Define seat bins and labels
bins = [0, 10, 30, 1000]
labels = ['Small (1-10 seats)', 'Medium (11-30 seats)', 'Large (31+ seats)']

# Create the seat segment category
accounts['seat_segment'] = pd.cut(accounts['seats'], bins=bins, labels=labels)

# Calculate churn rate per seat-size group
seat_churn = accounts.groupby('seat_segment', observed=False)['churn_flag'].agg(
    total_accounts='count',
    churned_accounts='sum',
    churn_rate='mean'
).reset_index()

# Convert to percentage and sort
seat_churn['churn_rate'] = seat_churn['churn_rate'] * 100
seat_churn = seat_churn.sort_values(by='churn_rate', ascending=False)

display(seat_churn)
```

# Bussiness Insight :- Small (1-10 seats ) customer seat-size groups churn the most.

# How much monthly and annual recurring revenue is at risk?


```python
# Calculate total MRR and ARR at risk (associated with churned accounts)
# We can sum these from the subscriptions of churned accounts

# Filter subscriptions for accounts that have churned
churned_account_ids = accounts[accounts['churn_flag'] == True]['account_id']
churn_subs = subscriptions[subscriptions['account_id'].isin(churned_account_ids)]

at_risk_mrr = churn_subs['mrr_amount'].sum()
at_risk_arr = churn_subs['arr_amount'].sum()

print(f"Total Monthly Recurring Revenue (MRR) at Risk: ${at_risk_mrr:,.2f}")
print(f"Total Annual Recurring Revenue (ARR) at Risk: ${at_risk_arr:,.2f}")
```

# Bussiness Insight :- Total Monthly Recurring Revenue (MRR) atRisk: $2,355,245.00 Total Annual Recurring Revenue (ARR) at Risk: $28,262,940.00 are monthly and annual recurring revenue is at risk.

### **Business Insight: Comparison of Active vs. Churned Customers**

Looking closely at the averages, we observe highly striking patterns:

* **Product Engagement is NOT the main driver of churn**:
  * Surprisingly, **Churned Customers** show almost identical levels of overall interaction with the product compared to Active Customers (`total_usage_count` of ~496 vs ~498, and `total_usage_duration_secs` of ~150.5k vs ~151.4k).
  * **Platform Stability is also comparable**: The error count encountered during usage (`total_usage_errors` of ~27.2 vs ~28.5) is slightly lower for churned accounts, meaning technical bugs are not disproportionately pushing them away.

* **Support Experience & Volume is Not the Differentiator**:
  * Both cohorts submit virtually the same number of tickets (~3.9 for Churned vs ~4.0 for Active) and experience almost identical average resolution times (~35 hours vs ~36 hours).
  * Customer satisfaction scores (`avg_satisfaction_score`) are also nearly matching (~3.98 for Churned vs ~3.95 for Active).

### **Strategic Takeaway**
Because raw usage quantities and standard support metrics (volume, resolution speeds, CSAT) are uniform across both groups, **churn is likely driven by qualitative issues** (e.g., missing specific high-value features, mismatch in pricing/budget, or changing business needs) rather than an overall lack of product adoption or poor support execution. This correlates strongly with our previous finding where "Features", "Budget", and "Pricing" ranked among the top reasons for churn.

# Which segments should be prioritised for retention?

```python
import pandas as pd

# Define segments to analyze
segments = ['industry', 'referral_source', 'plan_tier', 'seat_segment']

# Calculate and display churn metrics for each segment
for segment in segments:
    print(f"\n--- Churn Rate by {segment.replace('_', ' ').title()} ---")

    summary = accounts.groupby(segment, observed=False)['churn_flag'].agg(
        total_accounts='count',
        churned_accounts='sum',
        churn_rate='mean'
    ).reset_index()

    # Format as percentage and sort by highest churn rate
    summary['churn_rate_%'] = (summary['churn_rate'] * 100).round(2)
    summary = summary.drop(columns=['churn_rate']).sort_values(by='churn_rate_%', ascending=False)

    display(summary)
```

### **Business Insight: Prioritizing Customer Segments for Retention**

Based on the churn rate breakdown across different segments, we can pinpoint precisely where our retention efforts will yield the highest return:

1. **Primary Priority: Industry (DevTools)**
   * **Finding:** The **DevTools** industry has an exceptionally high churn rate of **30.97%** compared to other sectors like Cybersecurity (16.00%) or EdTech (16.46%).
   * **Action:** Investigate if there are product incompatibilities, integration needs, or competitive features missing specifically for technical/developer accounts.

2. **Secondary Priority: Acquisition Channel (Event Referrals)**
   * **Finding:** Customers acquired through **events** exhibit a **30.21%** churn rate—significantly higher than organic sign-ups (17.54%) or partner channels (14.61%).
   * **Action:** Review the qualification process for event-based leads. They might have lower intent or incorrect expectations about the product's value proposition.

3. **Tertiary Priority: Account Size (Small Accounts)**
   * **Finding:** **Small accounts (1-10 seats)** churn at **23.74%**, which is higher than medium (21.35%) and large (20.00%) accounts.
   * **Action:** Optimize onboarding or implement automated guidance for small teams to help them realize value quickly without requiring heavy manual account management.

4. **Tier Equality (Plan Tiers)**
   * **Finding:** Churn is almost perfectly uniform across all plan tiers (~22% for Enterprise, Pro, and Basic).
   * **Action:** Since plan tier does not significantly skew churn risk, prioritization should focus on **Industry** and **Referral Source** first.
